```python
# %%
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "visitor_features.parquet"

df = pd.read_parquet(DATA_PATH)
df.head()
```

```python
# %%
df["event_date"] = pd.to_datetime(df["visitor_first_seen_at_utc"]).dt.date
```

```python
# %%
daily_metrics = (
    df.groupby("event_date")
    .agg(
        visitors=("visitor_key", "nunique"),
        signups=("has_signup", "sum"),
        tests_started=("test_engagement_state", lambda x: (x != "No Test").sum()),
        tests_completed=("test_engagement_state", lambda x: (x == "Test Completed").sum()),
    )
    .reset_index()
)

daily_metrics.head()
```
